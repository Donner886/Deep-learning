```python
# -*- coding: utf-8 -*-
import numpy as np
import math
import time
```

```python
start_time = time.time()
print(f'starting at {start_time}')
# Create random input and output data
x = np.linspace(-math.pi, math.pi, 2000)
y = np.sin(x)

# Randomly initialize weights
a = np.random.randn()
b = np.random.randn()
c = np.random.randn()
d = np.random.randn()

learning_rate = 1e-6
for t in range(2000):
    # Forward pass: compute predicted y
    # y = a + b x + c x^2 + d x^3
    y_pred = a + b * x + c * x ** 2 + d * x ** 3

    # Compute and print loss
    """
        The loss is a sum over all samples, but the gradient is computed per sample and stored as a vector. 
        
        Deeper understanding of the gradient computation.
            
        
    """
    loss = np.square(y_pred - y).sum()
    if t % 100 == 99:
        print(t, loss)

#     # Backprop to compute gradients of a, b, c, d with    respect to loss
    grad_y_pred = 2.0 * (y_pred - y)
    grad_a = grad_y_pred.sum()
    grad_b = (grad_y_pred * x).sum()
    grad_c = (grad_y_pred * x ** 2).sum()
    grad_d = (grad_y_pred * x ** 3).sum()

#     # Update weights
    a -= learning_rate * grad_a
    b -= learning_rate * grad_b
    c -= learning_rate * grad_c
    d -= learning_rate * grad_d

print(f'Result: y = {a} + {b} x + {c} x^2 + {d} x^3')
end_time = time.time()
print(f'ending at {end_time}')
print(f'overall spent time: {end_time - start_time}')
```

### PyTorch: Tensors

A third order polynomial, trained to predict $ y = sin(x) $ from $ - pi $ to $ pi $ by minimizing squared Euclidean distance. 

This implementation uses PyTorch tensors and manually compute the forward pass, loss, and backward pass. 

A PyTorch Tensor is basically the same as a numpy array: it does not know anything about deep learning or computational graphs or gradients, and is just a generic n-dimensional array to be used for numerical computations.

The biggest difference between a numpy array and a PyTorch Tensor is that a PyTorch Tensor can run on either CPU or GPU. To run operations on the GPU, just cast the Tensor to a cuda datatype.




```python
import torch
import time
```

```python
start_time = time.time()
print(f'starting at {start_time}')
dtype = torch.float
device = torch.device('cpu')
#device = torch.device("cuda:0") # Uncomment this to run on GPU

# Create random input and output data
x = torch.linspace(-torch.pi, torch.pi, 2000, device=device, dtype=dtype)
y = torch.sin(x)


# Randomly initialize weights
a = torch.randn((), device=device, dtype=dtype)
b = torch.randn((), device=device, dtype=dtype)
c = torch.randn((), device=device, dtype=dtype)
d = torch.randn((), device=device, dtype=dtype)


earning_rate = 1e-6
for t in range(2000):
    # Forward pass: compute predicted y
    y_pred = a + b * x + c * x ** 2 + d * x ** 3

    # Compute and print loss
    loss = (y_pred - y).pow(2).sum().item()
    if t % 100 == 99:
        print(t, loss)

    # Backprop to compute gradients of a, b, c, d with respect to loss
    # grad_y_pred represents the partial derivative of the loss with respect to each individual prediction y^i
    grad_y_pred = 2.0 * (y_pred - y)
    grad_a = grad_y_pred.sum()
    grad_b = (grad_y_pred * x).sum()
    grad_c = (grad_y_pred * x ** 2).sum()
    grad_d = (grad_y_pred * x ** 3).sum()

    # Update weights using gradient descent
    a -= learning_rate * grad_a
    b -= learning_rate * grad_b
    c -= learning_rate * grad_c
    d -= learning_rate * grad_d


print(f'Result: y = {a.item()} + {b.item()} x + {c.item()} x^2 + {d.item()} x^3')
end_time = time.time()
print(f'ending at {end_time}')
print(f'overall spent time: {end_time - start_time}')   
```

### Why GPU initialisation can hang after a `fork()`

This is a very sharp and important question. Your intuition is correct: any process that accesses the GPU requires a valid execution context associated with its own PID. However, there is one key misconception: the GPU is not “locked to one PID” simply because it has been initialised once. In fact, multiple processes can each create their own GPU context.

The real problem is not that the GPU is exclusive by design; it is that `fork()` copies stale GPU state from the parent process, and the child process then tries to use that context with the wrong PID.

#### 1. Normal case: multiple processes can initialise the GPU safely
The GPU driver and runtime are designed to support multiple processes concurrently.

For example, if three independent Python processes initialise the GPU:

- Process A initialises the GPU → the driver creates context A for PID A
- Process B initialises the GPU → the driver creates context B for PID B
- Process C initialises the GPU → the driver creates context C for PID C

These contexts coexist independently in the driver. They do not permanently lock the GPU. As long as the required resources are available, they can share the device safely.

#### 2. The real problem: `fork()` copies the wrong GPU state
The issue arises when the parent process initialises the GPU before forking.

Suppose:
- Parent PID = 1000
- Child PID = 1001

The parent creates a GPU context and the driver records something like:
`{ channel: 0x99, bound_pid: 1000 }`

Then the parent calls `fork()`. Because `fork()` copies the parent memory image, the child inherits the same GPU context metadata:
`{ channel: 0x99, bound_pid: 1000 }`

Now the child tries to use that context, but the request is issued by PID 1001. The driver compares the request against the recorded binding and sees a mismatch:

- request from PID 1001
- context bound to PID 1000

The driver rejects the request as invalid. The child process then waits for hardware output that never arrives, which leads to a hang or deadlock.

#### Correct sequence: `fork()` first, initialise later
The safe pattern is:

1. Create the child process with `fork()`
2. Let each process initialise the GPU independently
3. Each process creates its own valid GPU context bound to its own PID

In that case:

- Parent PID 1000 creates context `{ channel: 0x77, bound_pid: 1000 }`
- Child PID 1001 creates context `{ channel: 0x88, bound_pid: 1001 }`

Now both processes have valid, independent GPU contexts and can run concurrently without conflicting PID bindings.

#### Final takeaway
The problem is not that “GPU initialisation locks the device.” The real issue is that the child process inherits a GPU context created for the parent, and the driver detects a PID mismatch. This is why the recommended pattern is:

- Correct: `fork()` first, then initialise the GPU
- Incorrect: initialise the GPU first, then `fork()`

This ensures each process obtains its own valid GPU context, and the driver can correctly match requests to the right PID.

```python
dtype = torch.float
device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else "cpu"
print(f"Using {device} device")
torch.set_default_device(device)
```

```python
## Create Tensors to hold input and outputs
## By default, requires_grad=False, which indicates that we do not need to compute gradients with respect to these Tensors during the backward pass.

x = torch.linspace(-1, 1, 2000, dtype=dtype)
y = torch.exp(x)   # A Taylor expansion would be 1 + x + (1/2) x**2 + (1/3!) x**3 + ...

# Create random Tensors for weights. For a third order polynomial, we need
# 4 weights: y = a + b x + c x^2 + d x^3
# Setting requires_grad=True indicates that we want to compute gradients with
# respect to these Tensors during the backward pass.
a = torch.randn((), dtype=dtype, requires_grad=True)
b = torch.randn((), dtype=dtype, requires_grad=True)
c = torch.randn((), dtype=dtype, requires_grad=True)
d = torch.randn((), dtype=dtype, requires_grad=True)

initial_loss = 1.
learning_rate = 1e-5

for t in range(5000):
    # Forward pass: compute predicted y using the operations on Tensor. 
    y_pred = a + b * x + c * x**2 + d * x**3
    # Compute and print loss using operations on Tensors.
    # Now loss is a Tensor of shape (1,)
    # loss.item() gets the scalar value held in the loss.
    loss = (y_pred - y).pow(2).sum()  # Sum of squared errors
    
    # Calculare initial loss, so we can report loss relative to it
    if t==0:
        initial_loss=loss.item()
    if t % 100 == 99:
        print(f'Iteration t = {t:4d}  loss(t)/loss(0) = {round(loss.item()/initial_loss, 6):10.6f}  a = {a.item():10.6f}  b = {b.item():10.6f}  c = {c.item():10.6f}  d = {d.item():10.6f}') 
    
    # Use autograd to compute the backward pass. This call will compute the
    # gradient of loss with respect to all Tensors with requires_grad=True.
    # After this call a.grad, b.grad. c.grad and d.grad will be Tensors holding
    # the gradient of the loss with respect to a, b, c, d respectively.
    loss.backward()
    if t % 100 == 99:
        print(f'a grad = {a.grad.item():10.6f}  b grad = {b.grad.item():10.6f}  c grad = {c.grad.item():10.6f}  d grad = {d.grad.item():10.6f}')
    # Manually update weights using gradient descent. Wrap in torch.no_grad()
    # because weights have requires_grad=True, but we don't need to track this
    # in autograd.
    with torch.no_grad():
        a -= learning_rate * a.grad
        b -= learning_rate * b.grad
        c -= learning_rate * c.grad
        d -= learning_rate * d.grad
        # Manually zero the gradients after updating weights
        a.grad = None
        b.grad = None
        c.grad = None
        d.grad = None   
print(f'Result: y = {a.item()} + {b.item()} x + {c.item()} x^2 + {d.item()} x^3')
```
